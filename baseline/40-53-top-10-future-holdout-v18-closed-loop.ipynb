# 40/53 Top-10 Future Holdout | v18 Closed Loop

**Public Kaggriculture agent | outcome-trained expert gate | both seats | pure-Python runtime | 2026-08-04 snapshot**

The **40/53** in the title is a chronological counterfactual replay holdout
against the current leaderboard top ten. It is **not** an official Public-LB
score. The official score is attached only after Kaggle evaluates the linked
Notebook submission.

v18 was built after v17 decayed from a strong historical validation result to
only **3/53** wins against newer top-team submissions. Instead of learning how
public agents order one pair of SELL actions, v18 optimizes complete-game wins
and chooses complete market policies containing SELL timing and quantity,
BUY, HIRE, and land decisions.

このNotebookは、良い結果だけでなく「閉ループ化でまだ増やせていない勝利」や
「検証で係数0になった盤面経路ゲート」も公開します。最後のセルは、Notebook
経由提出に使う自己完結`main.py`を完全復元します。

## 1. Why v17 became weak so quickly

The failure was not a mysterious neural-network bug. The leaderboard population
changed, while v17 kept yesterday's production trajectory. Its tiny MLP was also
too narrow: it could only permute existing non-WHEAT SELL orders and changed just
**7 turns (0.23% of actions)** in the diagnostic replay set.

| Agent on 2026-08-04 Top-10 snapshot | Older train | Later holdout |
|---|---:|---:|
| v17 learned SELL ranker | 5/49 | 3/53 |
| **v18 win-objective policy** | **44/49** | **40/53** |

Several current leaders differ from v17 in only 3–4 farmer/hand turns, but in
roughly 100+ market turns. That makes a market-policy refresh more important than
adding another threshold to the old route.

## 2. Policy: coherent board, closed-loop market

```text
public observation
   ├─ seat 0 / seat 1 win prior
   ├─ own money, hands, land, crops, animals, shed
   └─ nine normalized public market prices
                    │
                    ▼  once per day + hysteresis
       choose 1 of 4 complete market experts
       (SELL time/order/quantity + BUY + HIRE)

board route: selected per seat, kept coherent for the game
market route: state-selected at day boundaries, never every turn
```

There is no team-name lookup and no private opponent feature. The state vector
has 29 public/own features. Distance to each expert trajectory supplies the
closed-loop correction; log-smoothed episode wins supply separate seat priors.

The gate is deliberately slow-changing. On validation it selected a non-base
market expert on **1.1%** of daily decisions and switched on **0.6%**. One-turn
bait therefore cannot cause route churn.

````python
# Compact, readable form of the deployed decision rule.
from math import log

def expert_score(state, prototype, scale, seat_win_prior,
                 distance_strength=0.5, stay_bonus=0.5, is_current=False):
    distance = sum(
        ((value - center) / max(1e-12, spread)) ** 2
        for value, center, spread in zip(state, prototype, scale)
    ) / len(state)
    return seat_win_prior - distance_strength * distance + stay_bonus * is_current

print("Outcome prior + public-state distance + daily hysteresis")
````

## 3. The objective is winning, not action imitation

Every candidate is run through a complete 719-action game. Selection order is:

1. worst-seat win rate;
2. overall episode win rate;
3. smallest policy change that actually adapts;
4. switch rate;
5. coin margin only as a final tie-break.

This distinction matters because Kaggriculture rating reacts to wins and opponent
rating, not to an attractive offline action-label accuracy. v17's pairwise MLP
could classify order preferences without changing enough match outcomes. v18
trains its priors and hyperparameters from the result we actually want.

## 4. Walk-forward split: teams, time, and seats

No future replay is used to choose its own policy.

| Split | Contract | Games | Wins | Seat 0 | Seat 1 |
|---|---|---:|---:|---:|---:|
| fit | older episodes, first 6 teams | 11 | 9 base-route wins | 4/5 | 5/6 |
| validation | later episodes, same 6 teams | 12 | **11** | **6/6** | **5/6** |
| unseen | later episodes, disjoint 4 teams | 7 | **5** | **3/4** | **2/3** |
| full current holdout | all later Top-10 episodes | 53 | **40** | **21/27** | **19/26** |

The disjoint-team set is small, so 5/7 is evidence rather than a guarantee.
Replay opponents are open-loop and cannot react to our counterfactual actions;
official evaluation remains the final judge.

````python
# Reproducible presentation of the frozen evaluation report.
import pandas as pd
import matplotlib.pyplot as plt

rows = [
    {"agent": "v17", "train": 5/49, "future holdout": 3/53},
    {"agent": "v18 fixed", "train": 44/49, "future holdout": 40/53},
    {"agent": "v18 closed loop", "train": 44/49, "future holdout": 40/53},
]
frame = pd.DataFrame(rows).set_index("agent")
display((100 * frame).round(1).astype(str) + "%")
ax = (100 * frame).plot.bar(rot=0, figsize=(8, 4), color=["#94a3b8", "#0ea5e9"])
ax.set_ylabel("Win rate (%)")
ax.set_ylim(0, 100)
ax.set_title("Chronological Top-10 counterfactual replay evaluation")
ax.grid(axis="y", alpha=0.25)
plt.tight_layout()
````

## 5. Ablations that prevent decorative complexity

### Closed-loop market

The state gate changed a small number of daily expert choices, but **v18 fixed and
v18 closed loop both finished 40/53**. The gate preserved the downside in current
replays, but it has not yet created an additional win. The main recovery comes
from the refreshed complete policy, not from claiming that any adaptive model is
automatically better.

### Multiple board routes

The four current candidates share identical farmer/hand actions until turn 632.
v18 implements a safe one-time route gate exactly at that common-prefix fork and
locks the choice afterward. A nonzero gate selected another route in 3/12
validation games but remained 11/12. Two genuinely different v16 board routes
also produced the same 40/53 and the exact same loss set, with lower coin margin.

Therefore the published artifact keeps the route-gate structure but selects
`board_distance_strength = 0.0`. Route diversity is promoted only when it adds
wins, not because a portfolio sounds sophisticated.

## 6. What is learned and what remains fixed

**Learned from episode outcomes**

- separate seat-0 and seat-1 expert priors;
- complete market-expert preference;
- public-state distance coefficient;
- daily stay bonus;
- whether the safe board fork should activate.

**Kept deterministic for reliability**

- the selected farmer and hand trajectory between safe forks;
- standard-library-only inference;
- no runtime training, network access, LLM call, or opponent identity;
- exact fallback paths for v13–v17 ablations.

This is a lightweight outcome-trained mixture-of-experts gate—not a claim that
a large neural network can be learned reliably from a few dozen replay games.

## 8. Limitations and update rule

1. Current top policies can converge on the same public route within hours.
2. Counterfactual replay evaluation cannot model an opponent reacting to us.
3. The unseen-team sample is only seven games.
4. The market gate is safe in this snapshot but has not added a replay win.
5. Official strength is known only after the Notebook-linked submission runs.

The update rule is intentionally mechanical: refresh chronologically, preserve
team separation, evaluate both seats, optimize wins before margin, compare every
adaptive gate with its fixed ablation, and disable any component that does not
improve the win objective.
